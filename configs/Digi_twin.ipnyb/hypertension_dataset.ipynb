```python
from pathlib import Path
import numpy as np
import pandas as pd

DATA_DIR = Path.home() / "Downloads"
rng = np.random.default_rng(42)

diabetes_df = pd.read_csv(DATA_DIR / "diabetes_prediction_dataset.csv")
heart_df = pd.read_csv(DATA_DIR / "heart_2020_cleaned.csv")
print("diabetes:", diabetes_df.shape, "| heart:", heart_df.shape)

AGE_BINS = [17, 24, 29, 34, 39, 44, 49, 54, 59, 64, 69, 74, 79, 200]
AGE_LABELS = ["18-24", "25-29", "30-34", "35-39", "40-44", "45-49", "50-54",
              "55-59", "60-64", "65-69", "70-74", "75-79", "80 or older"]
GEN = {"Poor": 0, "Fair": 1, "Good": 2, "Very good": 3, "Excellent": 4}
yn = lambda s: (s == "Yes").astype(int)

base = diabetes_df[diabetes_df.gender.isin(["Female", "Male"]) & (diabetes_df.age >= 18)].copy()
base["sex"] = base.gender
base["age_band"] = pd.cut(np.floor(base.age), AGE_BINS, labels=AGE_LABELS).astype(str)
base["diab"] = base.diabetes.astype(int)
base["hd"] = base.heart_disease.astype(int)
base["bmi_key"] = base.bmi + rng.normal(0, 0.75, len(base))  


donor = pd.DataFrame({
    "sex": heart_df.Sex, "age_band": heart_df.AgeCategory,
    "diab": heart_df.Diabetic.str.startswith("Yes").astype(int),
    "hd": yn(heart_df.HeartDisease),
    "bmi_key": heart_df.BMI, "donor_bmi": heart_df.BMI,
    "physical_activity": yn(heart_df.PhysicalActivity), "alcohol": yn(heart_df.AlcoholDrinking),
    "stroke": yn(heart_df.Stroke), "kidney_disease": yn(heart_df.KidneyDisease),
    "asthma": yn(heart_df.Asthma), "diff_walking": yn(heart_df.DiffWalking),
    "bad_physical_days": heart_df.PhysicalHealth, "bad_mental_days": heart_df.MentalHealth,
    "gen_health": heart_df.GenHealth.map(GEN), "sleep_hours": heart_df.SleepTime,
    "skin_cancer": yn(heart_df.SkinCancer), 
    "bad_physical_days": heart_df.PhysicalHealth, "bad_mental_days": heart_df.MentalHealth,
    "gen_health": heart_df.GenHealth.map(GEN), "sleep_hours": heart_df.SleepTime,
}).sort_values("bmi_key")

merged = pd.merge_asof(base.sort_values("bmi_key"), donor, on="bmi_key",
                       by=["sex", "age_band", "diab", "hd"], direction="nearest")
n0 = len(merged)
merged = merged.dropna(subset=["gen_health"])
print(f"unmatched (no donor in block): {n0 - len(merged)} of {n0}")
merged["bmi_gap"] = (merged.bmi - merged.donor_bmi).abs()
print("median BMI gap:", round(merged.bmi_gap.median(), 2))
merged = merged.drop(columns=["sex", "age_band", "diab", "hd", "bmi_key", "donor_bmi"])
merged = merged.sample(frac=1, random_state=42).reset_index(drop=True)
merged.insert(0, "patient_id", merged.index)

merged.to_csv("merged_cohort.csv", index=False)
print(merged.shape, "->", Path("merged_cohort.csv").resolve())
print(merged.groupby("diabetes")[["gen_health", "bad_physical_days", "sleep_hours"]].mean().round(2))
merged.head()
```

```python
import pandas as pd
import numpy as np
from pathlib import Path

print("1. Loading clinical cohort...")
# Load the file you created earlier
clinical_df = pd.read_csv('merged_cohort.csv')

print("2. Loading and aggregating Fitbit data...")
fitbit_file = "/Users/apple/.cache/kagglehub/datasets/skywescar/lifesnaps-fitbit-dataset/versions/1/rais_anonymized/csv_rais_anonymized/daily_fitbit_sema_df_unprocessed.csv"
fitbit_df = pd.read_csv(fitbit_file, low_memory=False)

fitbit_features = [
    'nightly_temperature', 'nremhr', 'rmssd', 'spo2', 
    'full_sleep_breathing_rate', 'stress_score', 
    'sleep_points_percentage', 'TENSE/ANXIOUS', 'TIRED', 'ENTERTAINMENT'
]

fitbit_agg = fitbit_df.groupby('id')[fitbit_features].mean().reset_index()
fitbit_agg = fitbit_agg.rename(columns={'id': 'fitbit_id'})

print("3. Merging datasets...")
np.random.seed(42)
unique_ids = fitbit_agg['fitbit_id'].unique()
clinical_df['temp_fitbit_id'] = np.random.choice(unique_ids, size=len(clinical_df))

final_master = pd.merge(
    clinical_df, 
    fitbit_agg, 
    left_on='temp_fitbit_id', 
    right_on='fitbit_id', 
    how='left'
)

print("4. Cleaning up and adding patient_id...")
final_master = final_master.drop(columns=['temp_fitbit_id', 'fitbit_id'])

# CRITICAL: Add the patient_id column needed for the Streamlit Dashboard
if 'patient_id' not in final_master.columns:
    final_master.insert(0, 'patient_id', final_master.index)

print("5. Saving to CSV...")
output_filename = 'FINAL_dataset_with_fitbit.csv'
final_master.to_csv(output_filename, index=False)

print(f"✅ File successfully saved! Location: {Path(output_filename).resolve()}")
print(f"📊 Shape: {final_master.shape[0]} rows and {final_master.shape[1]} columns")
```
